## 🧠 El Primer Experimento Estadístico: Media vs. Mediana

Pregúntale a cualquier persona no técnica: *"¿Cuánto gasta en promedio un cliente en esta tienda?"*
La respuesta intuitiva siempre es: *"Calcula la Media aritmética"*.

Ejecuta la siguiente celda y observa los dos números.

```python
media_gasto = df['total_spend'].mean()
mediana_gasto = df['total_spend'].median()

print(f"Media Aritmética (Promedio): {media_gasto:.2f} €")
print(f"Mediana (El valor central real): {mediana_gasto:.2f} €")
print(f"Diferencia absoluta: {abs(media_gasto - mediana_gasto):.2f} €")
```
