## 🧹 Sección 2: Auditoría de Calidad y Detección de Nulos
En Pandas, combinamos `.isnull()` (máscara lógica) con `.sum()` para cuantificar los datos faltantes.

```python
# Contar valores nulos por columna de forma directa
print("Valores nulos por columna:")
print(df.isnull().sum())

# Porcentaje de nulos sobre el total
print("\nPorcentaje de nulos:")
print((df.isnull().mean() * 100).round(2))
```

## ⚖️ Sección 3: El Choque Estadístico (Media vs. Mediana)
*¿Cuánto gasta en promedio un cliente en esta tienda?*
- **Media Aritmética:** Altamente sensible a valores extremos (*Outliers*).
- **Mediana:** El percentil 50 real; insensible a compras multimillonarias.

```python
media_gasto = df['total_spend'].mean()
mediana_gasto = df['total_spend'].median()

print(f"Media Aritmética (Promedio): {media_gasto:.2f} €")
print(f"Mediana (El valor central real): {mediana_gasto:.2f} €")
print(f"Diferencia absoluta: {abs(media_gasto - mediana_gasto):.2f} €")
```

## 🧬 Sección 4: Anatomía de Pandas: Series y Frecuencias
Un `DataFrame` es una colección bidimensional de `pd.Series`. Analizamos las frecuencias univariantes con `.value_counts()`.

```python
# Gastos de la columna total_spend
gastos= df['total_spend']
print(type(gastos))


# Extraer los valores de la columna category
categorias=df['category'].value_counts().head(1)
print(categorias)
```

### 🏆 Identificación Automatizada del Líder (`.idxmax()` y `.max()`)
Buenas prácticas: calculamos el recuento una sola vez en memoria y extraemos tanto el nombre de la categoría como su valor máximo.

```python
conteo = df['category'].value_counts()  # Se calcula 1 sola vez en memoria
categoria_top = conteo.idxmax()
valor_top = conteo.max()
print(f"La categoría líder en ventas es {categoria_top} con un total de {valor_top} transacciones.")
```

## 🎯 Sección 5: Indexación Booleana (El `WHERE` de SQL en Pandas)
Filtramos filas evaluando máscaras lógicas vectorizadas sin necesidad de bucles.

### 💎 Caso 1: Detección de Compras VIP (`total_spend > 1000`)
Aislamos las transacciones de gran volumen y calculamos su gasto mediano real.

```python
compras_vip=df[df['total_spend']>1000]
# Cantidad de compras vip
total_con_shape=compras_vip.shape[0]
# Cantidad de compras vip con len()
total_con_len= len(compras_vip)
# Medias de las compras vip
media_clientes_vip =compras_vip['total_spend'].median()

print(f"\n Se han detectado {total_con_shape} transacciones VIP con un gasto mediano de {media_clientes_vip:.2f} €.")
```

### 📍 Caso 2: Filtros Múltiples con Operadores Bitwise (`&`)
Campaña exclusiva: Compras en **Madrid** con importe superior a **500 €**.

```python
madrid_premium= df[(df['city']=='Madrid') & (df['total_spend']>500)]
print(madrid_premium.head())
total_transacciones= madrid_premium.shape[0]
print(total_transacciones)

total_recaudo= madrid_premium['total_spend'].sum()


print(f"El total de transacciones mayores de 500 € en Madrid es de {total_transacciones} y se recaudo un total de {total_recaudo:.2f}€")
```

## 🌐 Sección 6: Agrupaciones y Agregaciones (El `GROUP BY` de SQL)
Aplicamos el patrón *Split-Apply-Combine* con `.groupby()` para consolidar métricas estratégicas por ciudad.

### 📊 Análisis Comparativo: La Paradoja de Madrid vs. Barcelona
¿Quién genera más y quién gasta más en promedio y mediana?

```python
df.groupby('city')['total_spend'].agg(['count', 'sum', 'median','mean']).sort_values(by='sum', ascending=False)
```

### 🎨 Presentación Ejecutiva con `display()` y Redondeo Financiero

```python

rendimiento_ciudades= df.groupby('city')['total_spend'].agg(['count', 'sum', 'mean']).sort_values(by='sum', ascending=False).round(2)
display(rendimiento_ciudades)
```
